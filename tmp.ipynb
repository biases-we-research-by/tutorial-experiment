```python
import csv
from tqdm import tqdm
myfile = open('enpages_parsed_models.csv', 'w')

writer = csv.DictWriter(myfile,fieldnames=['title','verbs','adjs','nouns'])
writer.writeheader()
for _,row in tqdm(df.iterrows(),total=len(df)):
    title = row['title']
    doc = nlp(row.text)
    verbs = list()
    adjs = list()
    nouns = list()

    for item in doc:
        if item.pos_ == 'VERB':
            verbs.append(item.lemma_.lower())
        elif item.pos_ == 'ADJ':
            adjs.append(item.lemma_.lower())
        elif item.pos_ == 'NOUN':
            nouns.append(item.lemma_.lower())
    writer.writerow({'title':title,'verbs':verbs,'adjs':adjs,'nouns':nouns})
```

```python
import glob,pandas as pd
from tqdm import tqdm


d = dict()

files = [x for x in glob.glob('output/**/P[0-9]*.csv')]
for doc in tqdm(files):
    df = pd.read_csv(doc)
    axis = doc.split('/')[-1].split('.')[0]
    if axis not in d:
        d[axis] = []
    if axis in ['P19','P27','P69']:
        for _,row in df.iterrows():
            try:
                d[axis].append(
                    {'entity':row.entity,
                    axis:row.unm49_mapping}
                )
            except:continue
    else:
        for _,row in df.iterrows():
            try:
                d[axis].append(
                    {'entity':row.entity,
                    axis:row[axis]}
                )
            except: continue

for k,v in d.items():
    pd.DataFrame(v).drop_duplicates().to_csv(f'tmp/{k}.csv',index=False)

```
