```python
# Cambia il valore di IsMain
import json

# Percorso del tuo file
file_path = '/Users/liadraetta/Desktop/progetti/Projects/tutorial-experiment/Data/raw data/teacher/teacher.json'

def update_json_file(path):
    # 1. Caricamento del file JSON
    try:
        with open(path, 'r', encoding='utf-8') as file:
            data = json.load(file)
    except FileNotFoundError:
        print(f"Errore: Il file {path} non esiste.")
        return

    # Se il file contiene una lista di oggetti, cicliamo su ognuno
    # Se è un oggetto singolo, lo trasformiamo in lista per uniformità
    items = data if isinstance(data, list) else [data]

    # 2. Modifica dei valori
    for item in items:
        claims = item.get("entity_claims", {})
        p106_values = claims.get("P106", [])

        # Verifica se il primo elemento è quello desiderato
        if p106_values and p106_values[0] == "Q1622272":
            item["isMain"] = True
        else:
            item["isMain"] = False

    # 3. Salvataggio del file aggiornato
    with open(path, 'w', encoding='utf-8') as file:
        # Se in origine era un oggetto singolo, salviamo il primo elemento
        # Altrimenti salviamo l'intera lista aggiornata
        json.dump(data, file, indent=4, ensure_ascii=False)
    
    print(f"File '{path}' aggiornato con successo!")

# Esegui la funzione
update_json_file(file_path)
```
